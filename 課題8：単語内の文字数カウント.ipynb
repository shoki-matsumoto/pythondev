```python
wordlist = []
checkdict = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"),0)
#カウント記録用辞書
checklist = list(checkdict.keys())
#辞書キーとの照合用リスト
while True:
    w = input("英単語を入力してください：")
    if w == (""):
        break

    for i in w:
        for j in range(len(checkdict)):
            if i == checklist[j]:
                checkdict[checklist[j]] += 1

    wordlist.append(w)

checkvalue = list(checkdict.values())
#辞書の値を取得

print(f"入力した英単語：{wordlist}")

for k in range(len(checklist)):
    if checkvalue[k] != 0:
        print(f"{checklist[k]}が{checkvalue[k]}個ありました")
```

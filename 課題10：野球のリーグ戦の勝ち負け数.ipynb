```python
class Baseball_Team:
    def __init__(self,name,win,lose,draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw
        self.calc_win_rate()

    def calc_win_rate(self):
        self.rate = round(self.win/(self.win + self.lose),3)

    def show_team_result(self):
        print(resultformat.format(self.name,self.win,self.lose,self.draw,self.rate))

headresultformat = "{:<9}{:>5}{:>6}{:>6}  {:<5}"
resultformat = "{:<9}{:>5}{:>6}{:>6}  {:<5.3f}"

Aruchers = Baseball_Team("Aruchers",77,64,2)
Meteors = Baseball_Team("Meteors",71,69,3)
Cheeatas = Baseball_Team("Cheetahs",69,68,6)
Sharks = Baseball_Team("Sharks",70,70,3)
pythons = Baseball_Team("pythons",68,73,2)
Owls = Baseball_Team("Owls",59,82,2)

print(headresultformat.format("team","win","lose","draw","rate"))
Aruchers.show_team_result()
Meteors.show_team_result()
Cheeatas.show_team_result()
Sharks.show_team_result()
pythons.show_team_result()
Owls.show_team_result()
```
